#### 2-D LINE PLOT
- Bivariate Analysis
- categorical -> numerical & numerical -> numerical
- Use case: Time series Data


```python
# plotting a simple function

price = [45000, 36000, 42000, 49000, 28000]
years = [2022, 2023, 2024, 2025, 2026]
plt.plot(years, price)
```

```python
# from a pandas dataframe

batsman = pd.read_csv('sharma-kohli.csv')
batsman

plt.plot(batsman['index'], batsman['V Kohli'])
```

```python
# plotting multiple plots

plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'], batsman['RG Sharma'])
```

```python
#labels title

plt.plot(batsman['index'],batsman['V Kohli'], color='green', linestyle='dashed', linewidth=3)
plt.plot(batsman['index'], batsman['RG Sharma'], color='black', linestyle='dashed', linewidth=2)

# linestyle - dashed / dotted /  dashdot ; color = hexcode of any color

plt.title("Virat Kohli Vs Rohit Sharma Career Comparison")
plt.xlabel('Years')
plt.ylabel('Runs Scored')
```

```python
plt.plot(batsman['index'],batsman['V Kohli'], color='green', linestyle='dashed', linewidth=3, marker='D', markersize=10, label='kohli')
plt.plot(batsman['index'], batsman['RG Sharma'], color='black', linestyle='dashed', linewidth=2, marker='o' ,markersize=8, label='Rohit')

# linestyle - dashed / dotted /  dashdot ; color = hexcode of any color

plt.title("Virat Kohli Vs Rohit Sharma Career Comparison")
plt.xlabel('Years')
plt.ylabel('Runs Scored')

plt.legend(loc = 'best')
```

```python
#limiting axes
price = [45000, 36000, 42000, 49000, 980000]
years = [2022, 2023, 2024, 2025, 2026]
plt.plot(years, price)

plt.ylim(0,100000)
plt.xlim(2022,2024)
```

```python
# grid & show

plt.plot(batsman['index'],batsman['V Kohli'], color='green', linestyle='dashed', linewidth=3, marker='D', markersize=10, label='kohli')
plt.plot(batsman['index'], batsman['RG Sharma'], color='black', linestyle='dashed', linewidth=2, marker='o' ,markersize=8, label='Rohit')
# linestyle - dashed / dotted /  dashdot ; color = hexcode of any color
plt.title("Virat Kohli Vs Rohit Sharma Career Comparison")
plt.xlabel('Years')
plt.ylabel('Runs Scored')
plt.legend(loc = 'best')

plt.grid()
plt.show()
```

#### Scatter Plots

- Bivariate Analysis
- numerical vs numerical
- Use cae: Finding Correlation

```python
#plt.scatter simple function
x = np.linspace(-10, 10, 50)
y = 10*x +3 + np.random.randint(0, 300, 50)

plt.scatter(x,y)
```

```python
#plt.scatter on pandas data

batter = pd.read_csv('batter.csv')
batter = batter.head(50)
batter

plt.scatter(batter['avg'], batter['strike_rate'], marker =  '+')
plt.title("Avg Vs Player's strike rate")
plt.xlabel('Average')
plt.ylabel('SR')
```

```python
# size
tips = sns.load_dataset('tips')
tips
plt.scatter(tips['total_bill'], tips['tip'], s=tips['size']*20)
```
