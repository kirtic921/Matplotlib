#### Pie Charts
- Univariate/ bivariate Analysis
- Categorical Vs NUmerical 
- Use case: To find a contribution on a standard scale

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Simple data

data = [10, 20, 30, 40, 50]
subjects = ['eng', 'hindi', 'sci', 'sst', 'maths']
plt.pie(data, labels=subjects)
plt.show()
```

```python
# dataset
df = pd.read_csv('gayle-175.csv')
df
```

```python
plt.pie(df['batsman_runs'], labels=df['batsman']) #, colors= ['Blue', 'Green', etc.]
plt.show()
```

```python
# explode window
plt.pie(df['batsman_runs'], labels=df['batsman'], autopct='%0.1f%%' ,explode=[0,0.1,0,0,0,0], shadow=True)
plt.show() 

plt.savefig('sample.png')
```
