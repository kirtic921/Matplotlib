#### Bar Chart

- Bivariate Analysis
- Numerical Vs Categorical
- Use case: Aggregate analysis of groups

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# simple bar chart
children = [10, 20, 30, 10, 15]
color = ['red', 'pink', 'green', 'yellow', 'blue']

plt.bar(color, children, color='green')
```

```python
# horizontal bar graph

plt.barh(color, children)
```

```python
# bar charts using data -> labels, colors
df = pd.read_csv('batsman_season_record.csv')
df
plt.bar(df['batsman'], df['2015'])
```

```python
df
```

```python
# Multiple Bar charts

plt.bar(np.arange(df.shape[0])-0.2, df['2015'], width=0.2)
plt.bar(np.arange(df.shape[0]), df['2016'], width=0.2)
plt.bar(np.arange(df.shape[0])+0.2, df['2017'], width=0.2)
```

```python
# xticks

plt.bar(np.arange(df.shape[0])-0.2, df['2015'], width=0.2)
plt.bar(np.arange(df.shape[0]), df['2016'], width=0.2)
plt.bar(np.arange(df.shape[0])+0.2, df['2017'], width=0.2)

plt.xticks(np.arange(df.shape[0]), df['batsman'])
plt.show()
```

```python
# overlapping

children = [10, 20, 30, 10, 15]
color = ['redredredredredredredred', 'pinkpinkpinkpinkpinkpink', 'greengreengreengreengreengreen', 'yellowyellowyellowyellowyellow', 'blueblueblueblueblue']

plt.bar(color, children)
plt.xticks(rotation = 'vertical')
plt.show()
```

```python
# Stacked Bar Chart

plt.bar(df['batsman'], df['2017'], label ='2017')
plt.bar(df['batsman'], df['2016'], bottom=df['2017'], label ='2016')
plt.bar(df['batsman'], df['2015'], bottom=(df['2016']+df['2017']), label ='2015')
plt.show()
```
